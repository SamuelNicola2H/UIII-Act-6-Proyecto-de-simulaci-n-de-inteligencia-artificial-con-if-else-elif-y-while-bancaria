```python
# ====================================================
# 🤖 IA DE CONCILIACIÓN BANCARIA 🤖
# ====================================================

print("==============================================")
print("🤖 IA DE CONCILIACIÓN BANCARIA 🤖")
print("==============================================")

continuar = "si"

while continuar == "si":

    saldo_libros = float(input("Ingresa el saldo en libros: "))
    saldo_banco = float(input("Ingresa el saldo del banco: "))

    print("")
    print("1. Cheque pendiente")
    print("2. Depósito no registrado")
    print("3. Comisión bancaria")
    print("4. Error de captura")
    print("5. Sin diferencias")

    opcion = int(input("Selecciona una opción: "))

    diferencia = saldo_libros - saldo_banco

    print("")
    print("🤖 Analizando información...")
    print("")

    if opcion == 1:

        print("📌 Posible causa: Cheque pendiente")

        if diferencia > 0:
            print("✅ Conciliación correcta")
        elif diferencia < 0:
            print("⚠️ Revisar cheque")
        else:
            print("✅ No hay diferencias")

    elif opcion == 2:

        print("📌 Posible causa: Depósito no registrado")

        if diferencia < 0:
            print("✅ Falta registrar depósito")
        else:
            print("⚠️ Revisar movimientos")

    elif opcion == 3:

        print("📌 Posible causa: Comisión bancaria")

        if diferencia > 0:
            print("💳 El banco descontó comisión")
        else:
            print("⚠️ Revisar estado de cuenta")

    elif opcion == 4:

        print("📌 Posible causa: Error de captura")

        if saldo_libros > saldo_banco:
            print("✏️ Error en libros")
        elif saldo_banco > saldo_libros:
            print("✏️ Error bancario")
        else:
            print("✅ Sin errores")

    elif opcion == 5:

        if saldo_libros == saldo_banco:
            print("🎉 Conciliación correcta")
        else:
            print("⚠️ Existen diferencias")

    else:
        print("❌ Opción inválida")

    print("")
    print("Saldo en libros:", saldo_libros)
    print("Saldo banco:", saldo_banco)
    print("Diferencia:", diferencia)

    continuar = input("\n¿Deseas realizar otra consulta? (si/no): ")

print("")
print("👋 Programa finalizado")
print("programa realizado por Samuel Nicola NL = 44")
```
