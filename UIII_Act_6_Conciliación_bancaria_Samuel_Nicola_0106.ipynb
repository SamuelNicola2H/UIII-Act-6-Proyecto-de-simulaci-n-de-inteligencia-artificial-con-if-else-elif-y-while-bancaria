## Conciliación bancaria


```python
# SIMULADOR DE INTELIGENCIA ARTIFICIAL PARA CONCILIACIÓN BANCARIA
# Nivel: Principiante (Sin funciones, sin POO, sin ciclos for, sin try)

print("====================================================")
print("🤖 IA DE CONCILIACIÓN BANCARIA - ASISTENTE VIRTUAL 🤖")
print("====================================================")
print("Bienvenido. Te ayudaré a revisar por qué el saldo de tu")
print("estado de cuenta bancario no coincide con tu libro contable.")
print("----------------------------------------------------\n")

# Solicitamos los datos iniciales para el análisis
saldo_banco = float(input("💰 Introduce el saldo final del ESTADO DE CUENTA BANCARIO: "))
saldo_libro = float(input("📖 Introduce el saldo final de tu LIBRO CONTABLE: "))

# El bucle principal mantendrá el programa activo hasta que los saldos cuadren
ejecutando = True

while ejecutando:
    print("\n----------------------------------------------------")
    print(f"DIFERENCIA ACTUAL: {saldo_banco - saldo_libro}")
    print(f"Saldo Banco: {saldo_banco} | Saldo Libro: {saldo_libro}")
    print("----------------------------------------------------")

    # Comprobación automática de la "IA"
    if saldo_banco == saldo_libro:
        print("\n🎉 ¡Felicidades! Los saldos coinciden perfectamente.")
        print("La conciliación bancaria se ha completado con éxito.")
        ejecutando = False # Termina el programa
    else:
        print("🤖 Análisis de IA: Detecto que los saldos NO coinciden.")
        print("Elige una de las siguientes opciones para registrar un ajuste:")
        print("1. Registrar un 'Cheque en Tránsito' (Emitido por ti, pero no cobrado en el banco aún).")
        print("2. Registrar un 'Depósito en Tránsito' (Enviado al banco, pero no aparece aún).")
        print("3. Registrar un 'Cargo Bancario' (Comisiones o intereses que el banco te cobró y no sabías).")
        print("4. Salir del asistente sin terminar.")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == "1":
            print("\n-> Has seleccionado: Cheque en Tránsito.")
            print("Explicación: Este dinero ya lo restaste de tu libro, pero el banco no lo sabe.")
            monto = float(input("Introduce el monto del cheque: "))
            # Para cuadrar, restamos el dinero del saldo del banco en la simulación
            saldo_banco = saldo_banco - monto
            print("🤖 IA: Ajuste aplicado al saldo del banco.")

        elif opcion == "2":
            print("\n-> Has seleccionado: Depósito en Tránsito.")
            print("Explicación: Este dinero ya lo sumaste en tu libro, pero el banco aún no lo procesa.")
            monto = float(input("Introduce el monto del depósito: "))
            # Para cuadrar, sumamos el dinero al saldo del banco en la simulación
            saldo_banco = saldo_banco + monto
            print("🤖 IA: Ajuste aplicado al saldo del banco.")

        elif opcion == "3":
            print("\n-> Has seleccionado: Cargo Bancario.")
            print("Explicación: El banco te quitó dinero que tú no habías registrado en tus libros.")
            monto = float(input("Introduce el monto de la comisión/cargo: "))
            # Para cuadrar, restamos el dinero de tu libro contable
            saldo_libro = saldo_libro - monto
            print("🤖 IA: Ajuste aplicado al saldo de tus libros.")

        elif opcion == "4":
            print("\n👋 Saliendo del asistente. Los saldos no quedaron conciliados.")
            ejecutando = False

        else:
            print("\n❌ Opción no válida. Por favor, selecciona un número del 1 al 4.")

print("\n====================================================")
print("🤖 Gracias por usar el Asistente de Conciliación Bancaria.")
print("🤖 programa elaborado por Samuel Nicola  NL = 44 👋")
print("====================================================")
```
